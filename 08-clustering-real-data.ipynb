## Part 1: Clustering arrests by U.S. state (25 points)

The dataset below contains statistics for the 50 U.S. states on four variables recorded in 1973:

- Murder: Arrests for murder (per 100,000 residents)
- Assault: Arrests for assault (per 100,000 residents)
- UrbanPop: Percentage of urban population
- Rape: Arrests for rape (per 100,000 residents)

1. Perform hierarchical clustering using complete linkage and Euclidean distance. Plot the resulting dendrogram.

2. Cut the dendrogram at a height that produces three clusters. Which states belong to each cluster?

3. Repeat the hierarchical clustering, again using complete linkage and Euclidean distance, after scaling all variables to have a standard deviation equal to one.

4. Compare the two clustering solutions. What effect does scaling the variables have on the resulting dendrogram and cluster memberships?
In your opinion, should the variables be scaled in this case before computing inter-observation dissimilarities? Provide a justification for your answer.

```python
import pandas as pd 
dat_1 = pd.read_csv("Lab08-part1_data.csv")
states = dat_1.state.to_list()
X = dat_1.drop(columns="state")
X.head() # This is the dataset to use
```

```python
# USE AS MANY CELLS AS NEEDED
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
import matplotlib.pyplot as plt

# Build the merge tree: complete linkage, Euclidean distance
Z_raw = linkage(X, method="complete", metric="euclidean")

plt.figure(figsize=(14, 5))
dendrogram(Z_raw, labels=states, leaf_rotation=90, leaf_font_size=8)
plt.title("Hierarchical Clustering (Complete Linkage, Unscaled Data)")
plt.ylabel("Euclidean distance")
plt.tight_layout()
plt.show()
```

Q1. The dendrogram above shows hierarchical clustering of the 50 states using complete linkage and Euclidean distance on the unscaled data. The highest merge occurs at a height of about 293, separating a group of high-assault states (e.g., FL, CA, NC) from the rest.

```python
# Cut the tree so we get exactly 3 clusters
clusters_raw = fcluster(Z_raw, t=3, criterion="maxclust")

result_raw = pd.DataFrame({"state": states, "cluster": clusters_raw})
for c in sorted(result_raw.cluster.unique()):
    members = result_raw[result_raw.cluster == c].state.to_list()
    print(f"Cluster {c} ({len(members)} states): {', '.join(members)}")
```

Q2. Cutting the dendrogram at a height of about 150 gives three clusters. Cluster 1 (16): AL, AK, AZ, CA, DE, FL, IL, LA, MD, MI, MS, NV, NM, NY, NC, SC. Cluster 2 (14): AR, CO, GA, MA, MO, NJ, OK, OR, RI, TN, TX, VA, WA, WY. Cluster 3 (20): CT, HI, ID, IN, IA, KS, KY, ME, MN, MT, NE, NH, ND, OH, PA, SD, UT, VT, WV, WI.

```python
X.groupby(clusters_raw).mean().round(1)
```

```python
from sklearn.preprocessing import StandardScaler

# Scale every column to mean 0 and standard deviation 1
X_scaled = StandardScaler().fit_transform(X)

Z_scaled = linkage(X_scaled, method="complete", metric="euclidean")

plt.figure(figsize=(14, 5))
dendrogram(Z_scaled, labels=states, leaf_rotation=90, leaf_font_size=8)
plt.title("Hierarchical Clustering (Complete Linkage, Scaled Data)")
plt.ylabel("Euclidean distance")
plt.tight_layout()
plt.show()

clusters_scaled = fcluster(Z_scaled, t=3, criterion="maxclust")
for c in sorted(set(clusters_scaled)):
    members = [s for s, k in zip(states, clusters_scaled) if k == c]
    print(f"\nCluster {c} ({len(members)} states):\n  " + "\n  ".join(members))
```

```python
pd.crosstab(clusters_raw, clusters_scaled,
            rownames=["Unscaled"], colnames=["Scaled"])
```

Q4. Scaling changes the results substantially. Without scaling, the dendrogram's height reaches about 293 and the three clusters are essentially a ranking by Assault (cluster means of about 273, 173, and 88), because Assault's values are far larger than the other variables' and dominate the Euclidean distance. UrbanPop barely differs across those clusters. After scaling, all four variables contribute equally: the high-crime states split into an urban group (e.g., CA, NY, IL, TX) and a less-urban Southern group (e.g., AL, MS, GA, the Carolinas), while the low- and moderate-crime states merge into one large cluster.

Yes, the variables should be scaled here. They are measured in different units (arrests per 100,000 vs. percent urban) and have very different variances, so without scaling the clustering mostly reflects Assault alone. Scaling ensures each variable contributes equally to the dissimilarities, which gives a more meaningful grouping of the states.
