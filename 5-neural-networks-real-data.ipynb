### Model setup

We didn't tune the model, we just picked common settings. The network has 2 hidden layers with 64 units each and ReLU, and one output unit with no activation since this is a regression problem. We used `nn.MSELoss` as the cost function, Adam with a learning rate of 0.001 (its default), a batch size of 32 and 300 epochs. To evaluate it we used the same 5-fold cross-validation from Section 1, and in every fold the scaler is fit only on the training part.

```python
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader
from sklearn.model_selection import KFold
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score


class ConcreteNN(nn.Module):
    def __init__(self, n_inputs, n_hidden):
        super().__init__()
        self.hidden1 = nn.Linear(n_inputs, n_hidden)
        self.hidden2 = nn.Linear(n_hidden, n_hidden)
        self.output = nn.Linear(n_hidden, 1)
        self.relu = nn.ReLU()

    def forward(self, x):
        x = self.relu(self.hidden1(x))
        x = self.relu(self.hidden2(x))
        return self.output(x)


print(ConcreteNN(n_inputs=8, n_hidden=64))
```

```python
def train_nn(X_train, y_train, epochs=300, lr=0.001, batch_size=32):
    model = ConcreteNN(n_inputs=X_train.shape[1], n_hidden=64)
    loss_fn = nn.MSELoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)
    loader = DataLoader(TensorDataset(X_train, y_train), batch_size=batch_size, shuffle=True)

    losses = []
    for epoch in range(epochs):
        epoch_loss = 0
        for X_batch, y_batch in loader:
            optimizer.zero_grad()
            loss = loss_fn(model(X_batch), y_batch)
            loss.backward()
            optimizer.step()
            epoch_loss += loss.item() * len(X_batch)
        losses.append(epoch_loss / len(X_train)) 
    return model, losses
```

### Training and evaluation with 5-fold cross-validation

This takes about a minute to run.

```python
kf = KFold(n_splits=5, shuffle=True, random_state=42)

fold_results = []
fold_losses = []
for fold, (train_idx, test_idx) in enumerate(kf.split(X), start=1):
    scaler = StandardScaler()
    X_train = scaler.fit_transform(X.iloc[train_idx])
    X_test = scaler.transform(X.iloc[test_idx])
    y_train = y.iloc[train_idx].values
    y_test = y.iloc[test_idx].values

    X_train = torch.tensor(X_train, dtype=torch.float32)
    X_test = torch.tensor(X_test, dtype=torch.float32)
    y_train = torch.tensor(y_train, dtype=torch.float32).reshape(-1, 1)

    torch.manual_seed(42)  
    model, losses = train_nn(X_train, y_train)
    fold_losses.append(losses)

    model.eval()
    with torch.no_grad():
        train_pred = model(X_train).numpy().ravel()
        test_pred = model(X_test).numpy().ravel()

    test_mse = mean_squared_error(y_test, test_pred)
    fold_results.append({
        "fold": fold,
        "train RMSE": mean_squared_error(y_train.numpy().ravel(), train_pred) ** 0.5,
        "test MSE": test_mse,
        "test RMSE": test_mse ** 0.5,
        "test R2": r2_score(y_test, test_pred),
    })

nn_results = pd.DataFrame(fold_results).set_index("fold")
pd.concat([nn_results, nn_results.agg(["mean", "std"])]).round(3)
```

```python
plt.figure(figsize=(8, 5))
for fold, losses in enumerate(fold_losses, start=1):
    plt.plot(losses, label=f"fold {fold}")
plt.yscale("log")
plt.xlabel("epoch")
plt.ylabel("training MSE (log scale)")
plt.title("Training loss for each fold")
plt.legend()
plt.show()
```

### Results

Across the 5 folds, the neural network had an average test RMSE of 5.36 MPa (standard deviation 0.36) and an average test R² of 0.896. This means a typical prediction on concrete the model hasn't seen is off by about 5.4 MPa, and the model explains about 90% of the variation in strength. That is a lot better than just guessing the average strength every time, which would give an RMSE of about 16.7 MPa (the standard deviation of strength from Section 2). The results were also pretty consistent across the folds, with the test RMSE only going from 4.92 to 5.78 MPa, so the model doesn't depend much on which rows end up in the test fold.

The training RMSE (3.61 MPa on average) is lower than the test RMSE (5.36 MPa), so the model fits the data it was trained on better than new data, which means there is some overfitting. The training loss curves drop really fast in the first few epochs, go from about 200 to about 30 by epoch 100, and then keep going down slowly until epoch 300. All 5 folds look almost the same, so the training was stable no matter which part of the data was used.

### Model setup

The scikit-learn neural network page has three models. We picked MLPRegressor because we are predicting a number (concrete strength). MLPClassifier is for predicting categories, and BernoulliRBM doesn't make predictions, so neither one works here.

We gave it the same setup as our PyTorch network:
2 hidden layers with 64 units each, using ReLU
an output layer with no activation, so it can predict any number
squared error as the loss
Adam with a learning rate of 0.001
batch size of 32 and 300 epochs

We didn't tune anything. We just copied the settings from our PyTorch model.

By default, scikit-learn adds a penalty on large weights and stops training early if the loss stops improving. Our PyTorch model doesn't do either, so we turned both off (alpha=0 and n_iter_no_change=300).

We used the same 5 folds as before. The scaler and the model are in a `Pipeline`, so the scaler only learns from the training part of each fold and never sees the test part.

```python
import warnings
from sklearn.neural_network import MLPRegressor
from sklearn.pipeline import make_pipeline
from sklearn.exceptions import ConvergenceWarning

sk_results = []
sk_losses = []
for fold, (train_idx, test_idx) in enumerate(kf.split(X), start=1):
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    model = make_pipeline(
        StandardScaler(),
        MLPRegressor(hidden_layer_sizes=(64, 64), activation="relu", solver="adam",
                     learning_rate_init=0.001, batch_size=32, max_iter=300,
                     alpha=0, n_iter_no_change=300, random_state=42),
    )
    # it warns that it hit max_iter, which is what we want here
    with warnings.catch_warnings():
        warnings.simplefilter("ignore", ConvergenceWarning)
        model.fit(X_train, y_train)

    # sklearn records half the MSE, so multiply by 2 to compare with nn.MSELoss
    sk_losses.append([2 * loss for loss in model[-1].loss_curve_])

    train_pred = model.predict(X_train)
    test_pred = model.predict(X_test)

    test_mse = mean_squared_error(y_test, test_pred)
    sk_results.append({
        "fold": fold,
        "train RMSE": mean_squared_error(y_train, train_pred) ** 0.5,
        "test MSE": test_mse,
        "test RMSE": test_mse ** 0.5,
        "test R2": r2_score(y_test, test_pred),
    })

sk_results = pd.DataFrame(sk_results).set_index("fold")
pd.concat([sk_results, sk_results.agg(["mean", "std"])]).round(3)
```

```python
plt.figure(figsize=(8, 5))
for fold, losses in enumerate(sk_losses, start=1):
    plt.plot(losses, label=f"fold {fold}")
plt.yscale("log")
plt.xlabel("epoch")
plt.ylabel("training MSE (log scale)")
plt.title("Training loss for each fold (scikit-learn MLP)")
plt.legend()
plt.show()
```

```python
comparison = pd.DataFrame({
    "PyTorch NN": nn_results.mean(),
    "sklearn MLP": sk_results.mean(),
}).T
comparison.round(3)
```

### Results

The two models did almost the same. The scikit-learn model had a test RMSE of 5.278 MPa and an R² of 0.899, and our PyTorch model had 5.326 MPa and 0.897. The difference is very small, so neither model is really better.

They aren't exactly the same because the two libraries start with different weights and shuffle the data in a different order. Since the results are so close, this shows our PyTorch network was built correctly.

```python
from sklearn.svm import SVR
from sklearn.linear_model import LinearRegression

def run_cv(make_model):
    results = []
    for fold, (train_idx, test_idx) in enumerate(kf.split(X), start=1):
        X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
        y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]


        model = make_model()
        model.fit(X_train, y_train)

        train_pred = model.predict(X_train)
        test_pred = model.predict(X_test)

        test_mse = mean_squared_error(y_test, test_pred)
        results.append({
            "fold": fold,
            "train RMSE": mean_squared_error(y_train, train_pred) ** 0.5,
            "test MSE": test_mse,
            "test RMSE": test_mse ** 0.5,
            "test R2": r2_score(y_test, test_pred),
        })
    return pd.DataFrame(results).set_index("fold")

svr_results = run_cv(lambda: make_pipeline(StandardScaler(), SVR(kernel="rbf")))
lr_results = run_cv(lambda: make_pipeline(StandardScaler(), LinearRegression()))

print("SVR Results")
display(pd.concat([svr_results, svr_results.agg(["mean", "std"])]).round(3))
print("Linear Regression Results")
display(pd.concat([lr_results, lr_results.agg(["mean", "std"])]).round(3))
```

```python
comparison = pd.DataFrame({
    "PyTorch NN": nn_results.mean(),
    "sklearn MLP": sk_results.mean(),
    "SVR": svr_results.mean(),
    "Linear Regression": lr_results.mean(),
}).T
comparison.round(3)
```
